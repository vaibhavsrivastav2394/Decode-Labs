```python
# PROJECT 2: EXPLORATORY DATA ANALYSIS (EDA) & DATA FORENSICS
import numpy as np
import pandas as pd

#DATA LOADING:CSV File
file_path = "Dataset_for_Data_Analytics.csv"

# 1 Load the dataset into a pandas DataFrame
df = pd.read_csv(file_path)

# Display initial rows to understand the raw data structure
print("--- [INPUT] FIRST 5 ROWS OF DATASET ---")
print(df.head())

# Check shape (Rows, Columns) and data types
print(f"\nDataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns")
print("\n--- DATA TYPES & NON-NULL COUNTS ---")
df.info()

# 2. PROCESS: Missing Values & Data Integrity

missing_counts = df.isnull().sum()
missing_percentage = (missing_counts / len(df)) * 100

missing_df = pd.DataFrame(
    {"Missing Values": missing_counts, "Percentage (%)": missing_percentage}
)
print("\n--- MISSING VALUE DIAGNOSIS (NOISE CHECK) ---")
print(missing_df[missing_df["Missing Values"] > 0])


df["CouponCode"] = df["CouponCode"].fillna("No_Coupon")


# 3. STATISTICS: count, mean, std, min, Q1, median (50%), Q3, max
print("\n--- DESCRIPTIVE STATISTICS (FIVE-NUMBER SUMMARY) ---")
print(df.describe())

# 4. OUTLIER
numeric_columns = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

print("\n--- OUTLIER DETECTION VIA IQR METHOD ---")
for col in numeric_columns:
  # Calculate Q1 (25th percentile) and Q3 (75th percentile) using NumPy/Pandas
  Q1 = np.percentile(df[col].dropna(), 25)
  Q3 = np.percentile(df[col].dropna(), 75)
  IQR = Q3 - Q1

  # Define lower and upper bounds
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  # Flag outliers (Suspects) using boolean indexing
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  print(
      f"Column '{col}': Found {len(outliers)} outliers | Safe Range:"
      f" [{lower_bound:.2f}, {upper_bound:.2f}]"
  )
```
